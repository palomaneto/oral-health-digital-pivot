# Exploratory Analysis: Oral Health Risk & Medical History

This notebook explores how patient medical history and demographic factors relate to oral health risk using a synthetic dataset.


## Project Objective

The objective of this analysis is to explore patterns between medical history and oral health risk to support preventive care planning.


## Dataset Overview

The dataset contains 200 synthetic patient records with the following fields:
- Patient ID
- Age
- Gender
- Medical history
- Oral health risk category
- Last dental visit date


## Key Questions

- Which medical conditions are most associated with high oral health risk?
- Are older patients more likely to be high risk?
- Are high-risk patients overdue for dental visits?


## Planned Analysis Method

SQL queries will be used to group patients by risk level, identify vulnerable populations, and summarise attendance patterns.
